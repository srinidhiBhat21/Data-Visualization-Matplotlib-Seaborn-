```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from warnings import filterwarnings
filterwarnings('ignore')
```

```python
df = sns.load_dataset("tips")
```

## Histogram (Matplotlib)

```python
plt.figure(figsize=(8,5))
plt.hist(df['total_bill'], bins = 20, color='skyblue', edgecolor='black')
plt.title('Distribution of Total Bills', fontsize = 16)
plt.xlabel('Total Bill', fontsize = 12)
plt.ylabel('Frequency', fontsize = 12)
plt.show()
```

## Seaborn

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='total_bill', bins=20, kde=True, color='skyblue')
plt.title('Distribution of Total Bills', fontsize=16)
plt.xlabel('Total Bill', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```

## Bar chart (Matplotlib)

```python
day_tips = df.groupby('day')['tip'].mean()
plt.figure(figsize=(8,5))
plt.bar(day_tips.index, day_tips.values, color='orange')
plt.title('Average Tip by Day', fontsize=16)
plt.xlabel('Day', fontsize=12)
plt.ylabel('Average Tip', fontsize=12)
plt.show()
```

## Seaborn

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='day', y='tip', palette='viridis')
plt.title('Average Tip by Day', fontsize=16)
plt.xlabel('Day', fontsize=12)
plt.ylabel('Average Tip', fontsize=12)
plt.show()
```

## BoxPlot(matplotlib)

```python
plt.figure(figsize=(8,5))
plt.boxplot(df['total_bill'])
plt.title('Box Plot of Total Bills', fontsize=16)
plt.ylabel('Total Bill', fontsize=12)
plt.show()
```

## seaborn

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='day', y='total_bill', palette='Set2')
plt.title('Total Bill Distribution by Day', fontsize=16)
plt.xlabel('Day', fontsize=12)
plt.ylabel('Total Bill', fontsize=12)
plt.show()
```

## Scatterplot(Matplotlib)

```python
plt.figure(figsize=(8,5))
plt.scatter(df['total_bill'], df['tip'], color='green', alpha=0.7)
plt.title('Total Bill vs Tip', fontsize=16)
plt.xlabel('Total Bill', fontsize=12)
plt.ylabel('Tip', fontsize=12)
plt.show()
```

## seaborn

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex', palette='Set1')
plt.title('Total Bill vs Tip by Gender', fontsize=16)
plt.xlabel('Total Bill', fontsize=12)
plt.ylabel('Tip', fontsize=12)
plt.show()
```

## Heatmap of correlations

```python
corr = df.select_dtypes(include='number').corr()
corr
```

## matplotlib

```python
plt.figure(figsize=(8,5))
plt.imshow(corr, cmap='coolwarm', interpolation='nearest')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Correlation Heatmap', fontsize=16)
plt.show()
```

## seaborn

```python
plt.figure(figsize=(8,5))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap', fontsize=16)
plt.show()
```

## Saving a chart as PNG

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='total_bill', kde=True, color='red')
plt.title('Distribution of Total Bills')
plt.savefig('total_bill_histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```
